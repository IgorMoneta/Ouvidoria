```
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity

df = pd.DataFrame(MANIFESTACOES, columns=["id","categoria","texto"])
modelo = SentenceTransformer("paraphrase-multilingual-MiniLM-L12-v2")
embeddings = modelo.encode(df.texto.tolist(), normalize_embeddings=True)
matriz = cosine_similarity(embeddings)

def detectar_duplicatas(textos, limiar=0.85):
    vetores = modelo.encode(textos, normalize_embeddings=True)
    sim = cosine_similarity(vetores)
    pares = []
    for i in range(len(textos)):
        for j in range(i+1, len(textos)):
            if sim[i,j] >= limiar:
                pares.append((i, j, float(sim[i,j])))
    return pares

pares = detectar_duplicatas(df.texto.tolist(), 0.85)
pares_formatados = pd.DataFrame([
    {"Documento A": df.iloc[i].id, "Documento B": df.iloc[j].id, "Similaridade": score}
    for i,j,score in pares
]).sort_values("Similaridade", ascending=False)
pares_formatados
```

```
plt.figure(figsize=(12,9))
sns.heatmap(matriz, xticklabels=df.id, yticklabels=df.id, cmap="viridis")
plt.title("Matriz de similaridade semântica")
plt.tight_layout()
plt.show()
```

## Justificativa do limiar

O limiar 0.85 é um ponto de partida conservador para marcar textos como duplicatas semânticas. A escolha deve ser validada contra as duplicatas reais do dataset: um limiar muito baixo aumenta falsos positivos, enquanto um limiar muito alto pode perder paráfrases legítimas. Para produção, é recomendável testar vários limiares e escolher aquele que oferece o melhor equilíbrio entre precisão e cobertura.

```
duplicatas_reais = {
    frozenset(("M003","M017")),
    frozenset(("M008","M022")),
    frozenset(("M004","M024")),
    frozenset(("M012","M027")),
    frozenset(("M005","M030")),
    frozenset(("M011","M032")),
}

encontradas = {frozenset((a,b)) for a,b,_ in [(df.iloc[i].id, df.iloc[j].id, s) for i,j,s in pares]}
tp = len(encontradas & duplicatas_reais)
fp = len(encontradas - duplicatas_reais)
fn = len(duplicatas_reais - encontradas)

analise = pd.DataFrame([{
    "Verdadeiros positivos": tp,
    "Falsos positivos": fp,
    "Falsos negativos": fn,
    "Precisão": tp / len(encontradas) if encontradas else 0,
    "Recall": tp / len(duplicatas_reais)
}])
analise
```
