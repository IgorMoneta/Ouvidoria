```
import numpy as np
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
from sentence_transformers import SentenceTransformer

df = pd.DataFrame(MANIFESTACOES, columns=["id","categoria","texto"])
textos = df["texto"].tolist()
indices = {x:i for i,x in enumerate(df.id)}

bow = CountVectorizer(lowercase=True, strip_accents="unicode")
X_bow = bow.fit_transform(textos)

tfidf = TfidfVectorizer(lowercase=True, strip_accents="unicode")
X_tfidf = tfidf.fit_transform(textos)

modelo = SentenceTransformer("paraphrase-multilingual-MiniLM-L12-v2")
X_emb = modelo.encode(textos, normalize_embeddings=True)

pares = [("M003","M017"),("M008","M022"),("M008","M031")]
linhas = []
for a,b in pares:
    i,j=indices[a],indices[b]
    linhas.append({
        "Par": f"{a} × {b}",
        "BoW": cosine_similarity(X_bow[i],X_bow[j])[0,0],
        "TF-IDF": cosine_similarity(X_tfidf[i],X_tfidf[j])[0,0],
        "Embeddings": cosine_similarity([X_emb[i]],[X_emb[j]])[0,0]
    })
comparacao = pd.DataFrame(linhas)
comparacao
```

## Conclusão

BoW e TF-IDF dependem principalmente da sobreposição literal de palavras. Por isso, frases que descrevem o mesmo problema com vocabulário diferente podem apresentar similaridade baixa. TF-IDF melhora a ponderação dos termos raros, mas continua sendo uma representação esparsa. Embeddings representam relações semânticas e contexto, sendo mais adequados para reconhecer paráfrases como “buraco enorme” e “asfalto esburacado” ou “posto sem médico” e “falta atendimento”. No par M008 × M031, que trata de saúde e iluminação, embeddings podem apresentar alguma proximidade temática por causa do contexto curto, mas devem manter score inferior ao de pares semanticamente equivalentes. Os valores exatos dependem do modelo e de sua versão.
